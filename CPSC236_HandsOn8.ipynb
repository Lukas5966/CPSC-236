## 10.1 Email Creator
Create a program that reads a file and creates a series of emails.

### Console:
```powershell
Email Creator

================================================================
To: jbutler@gmail.com 
From: noreply@deals.com 
Subject: Deals!

Hi James,

We've got some great deals for you. Check our website!
================================================================
To:	josephine_darakjy@darakjy.org 
From:	noreply@deals.com
Subject: Deals! 

Hi Josephine,

We've got some great deals for you. Check our website!
================================================================
To:	art@venere.org
From:	noreply@deals.com 
Subject: Deals!

Hi Art,

We've got some great deals for you. Check our website!
```

### Specifications:
- Use `email.csv` file:
- Use `email_template.txt` file:
- When the program starts, it should read the email addresses and first names from the file, merge them into the mass email template, and display the results on the console.
- All email addresses should be converted to **lowercase**.
- All first names should be converted to **title case**.
- If you add names to the list of email addresses, the program should create more emails.
- If you modify the template, the program should change the content of the email that’s created. 



````python
import csv
#open the email.csv file and create a list of lists#
#each list will contain information with the following indexes. [0] first name, [1] last name, [2] email address#
with open("email.csv", "r") as file:
    #use imported cvs pack to make a reader object#
    cvs = csv.reader(file)
    LS = []
    for row in cvs:
        LS.append(row)

#Template is stored in a string so the file can be opened once#
with open("email_template.txt", "r") as file:
        template = file.read()

#Start of primary output#
print("Email Creator\n")
for person in LS:
    #Crate variables for anything that needs replaced in the template (name and email), and change any capitals or lowercases that are needed#
    fName = person[0]
    fName = fName[:1].upper() + fName[1:]
    email_addr = person[2].lower()

    #generate email by storing the template in a string, then replacing {email} and {first_name}#
    email = template
    email = email.replace("{email}", email_addr)
    email = email.replace("{first_name}", fName)
    print("================================================================")
    print(email)
````

## 10.2 Pig Latin Translator

Create a program that translates English to Pig Latin.

### Console:
```powershell
Pig Latin Translator

Enter text: Tis but a scratch. 
English:	tis but a scratch
Pig Latin:	istay utbay away atchscray 

Continue? (y/n): y

Enter text: We are the knights who say nee!
English:	we are the knights who say nee
Pig Latin:	eway areway ethay ightsknay owhay aysay eenay 

Continue? (y/n): n

Bye!
```

### Specifications:
- Convert the English to lowercase before translating.
- Remove the following punctuation characters before translating `! , ; : - . ?`
- Assume that words are separated from each other by a single space.
- If the word starts with a vowel, just add *way* to the end of the word.
- If the word starts with a consonant, move all of the consonants that appear before the first vowel to the end of the word, then add *ay* to the end of the word.

### Note:
- There are no official rules for Pig Latin. Most people agree on how words that begin with consonants are translated, but there are many different ways to handle words that begin with vowels.


````python
print("Pig Latin Translator\n")
#Program will be within a while loop to allow the user to choose to do another or not#
while True:
    #Collect a string as input#
    s = input("Enter Text: ")
    #Remove punctuation characters#
    s = s.replace("!", "")
    s = s.replace(",", "")
    s = s.replace(";", "")
    s = s.replace(":", "")
    s = s.replace("-", "")
    s = s.replace(".", "")
    s = s.replace("?", "")
    s = s.lower()
    #Print plain text#
    print("English: " + s)
    #Split the string input into a list of words. The words will be translated before being printed#
    LS = s.split(" ")
    #Now that we have the individual words, we will use a for loop to translate each word#
    print("Pig Latin: ", end="")
    for word in LS:
        #create a string with the first letter of the word. This will be used to check if it a vowel, and also to move the letter to the end of the word#
        first = word[0]
        #if statement to check if it is a vowel#
        if (first == "a" or first == "e" or first == "i" or first == "o" or first == "u"):
            #do not move first letter, add "way" to the end of the word#
            word = word + "way"
        else:
            #move first letter to the end of the word, add "ay" to the end#
            word = word[1:] + first + "ay"
        #print the words seperated by a space to print the full phrase#
        print(word, end=" ")
    #Start a new line and ask if the user would like to continue#
    print("\n")
    if input("Continue? (y/n): ") == "n":
        print("Bye!")
        break
````
