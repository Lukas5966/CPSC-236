## 12.1 Champion Counter
Create a program that reads a text file that contains a list of FIFA World Cup champions and determines the country that has won the most championships.

### Console:
```powershell
Country	        Wins	Years	
=======	        ====	=====	
Argentina       2       1978, 1986
Brazil	        5       1958, 1962, 1970, 1994, 2002
England	        1       1966		
France	        1       1998		
Germany	        4       1954, 1974, 1990, 2014
Italy	        4       1934, 1938, 1982, 2006
Spain	        1       2010
Uruguay	        2       1930, 1950
```

### Specifications:
- Use the attached text file named `world_cup_champions.txt` that contains data like this: 
```powershell
Year,Country,Coach,Captain 
1930,Uruguay,Alberto Suppici,José Nasazzi 
1934,Italy,Vittorio Pozzo,Gianpiero Combi 
1938,Italy,Vittorio Pozzo,Giuseppe Meazza
...
```
- When the program starts, it should read the text file and use a dictionary to store the required data using the name of each country that has won the World Cup as the key.
- The program should compile the data shown above and display the countries alphabetically.



````python
with open('world_cup_champions.txt', 'r') as file:
    #Use readline to skip the labeling line#
    file.readline()
    
    #Take only needed information from file (Country and year). The country is the key, the item is a list of years the country has won#
    D = {}
    for x in file:
        LS = x.split(',')
        if LS[1] in D:
            D[LS[1]].append(LS[0])
        else:
            D[LS[1]] = [LS[0]]

#Okay we have a dictionary with countries as the keys and the corresponding list is each year they won#
#Now I will write the code to output the stored dictionary in the proper format#

#Header#
print("Country", "\t", "Wins", "\t", "Years")
#Seperating Line#
print("=======", "\t", "====", "\t", "=====")

#Information output#
#For loop to display each Key, .len() will be used to display number of wins, and each year will be displayed with another loop#
for k in D:
    print(k, "\t", end="")
    #Formatting so all the numbers stay in line#
    if len(k)<7:
        print("\t", end="")
    print(len(D[k]), "\t ", end="")
    for date in D[k]:
        if date != D[k][len(D[k]) - 1]:
            print(date, ", ", sep="", end="")
        else:
            print(date)
    
    
````

## 12.2 Monthly Sales
Create a program that allows you to view and edit the sales amounts for each month of the current year.

### Console:
```powershell
Monthly Sales program

COMMAND MENU
view	- View sales for specified month 
edit	- Edit sales for specified month 
totals - View sales summary for year 
exit	- Exit program

Command: view
Three-letter Month: jan
Sales amount for Jan is 14,317.00.

Command: edit
Three-letter Month: jan 
Sales Amount: 15293
Sales amount for Jan is 15,293.00.

Command: totals
Yearly total: 67,855.00
Monthly average: 5,654.58

Command: view
Three-letter Month: july 
Invalid three-letter month.

Command: exit Bye!
```

### Specifications:
- Use the attached text file named `monthly_sales.txt` that consists of rows that contain three-letter abbreviations for the month and the monthly sales.
- The program should read the file and store the sales data for each month in a dictionary with the month abbreviation as the key for each item.
- Whenever the sales data is edited, the program should write the changed data to the text file.




````python
with open("monthly_sales.txt", "r") as file:
    #This program will put all month abbreviations as keys into a dictionary with the number of sales as the corresponding item#
    #The user interface will allow the user to edit the dictionary, which will later overwrite the file after the edit command#
    D = {}
    for line in file:
        s = line
        s = s.replace("\n", "")
        s = s.lower()
        LS = s.split("\t")

        D[LS[0]] = LS[1]
    #Dictionary is created. The file can be closed for now, but will be reopened to write on later#

#User interface#
#Title and Command List#
print("Monthly Sale Program\n")
print("COMMAND MENU")
print("view\t- View sales for specified month")
print("edit\t- Edit sales for specified month")
print("totals\t- View sales summary for year")
print("exit\t- Exit program\n")

#Collect user input. Use a while loop to collect input until exit command is used#
while True:
    #Variables used in commands. They should be reset with each loop to ensure input is not taken from a previous command#
    cmd = input("Command: ")
    month = ""
    total = 0
    num = 0
    
    if cmd.lower() == "view":
        month = input("Three-Letter Month: ")
        if month.lower() in D:
            print("Sales amount for", month, "is", D[month.lower()], "\n")
        else:
            print("Selected abbreviation does not exist.")
    elif cmd.lower() == "edit":
        #Edit the dictionary, then open the file to write and write new dictionary into file#
        #careful to format it properly so file can be read again#
        month = input("Three-Letter Month: ")
        if month.lower() in D:
            num = input("Sales Amount: ")
            D[month.lower()] = num
            print("Sales Amount for", month, "is", D[month.lower()], "\n")

            #Now we open the file again to write the new information#
            with open("monthly_sales.txt", "w") as file:
                for key in D:
                    s = key + "\t" + D[key] + "\n"
                    file.write(s)
        else:
            print("Selected abbreviation does not exist.")
        
    elif cmd.lower() == "totals":
        for key in D:
            total = total + int(D[key])
        print("Yearly Totals:", total)
        print("Monthly Average:", (total/12), "\n")
    elif cmd.lower() == "exit":
        print("Bye!")
        break
    else:
        print("Invalid Input. Try again.\n")
````
